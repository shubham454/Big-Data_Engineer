```sql
-- Write a Query to provide the date for nth occurrence of Sunday in future from given datae
-- datepart
-- sunday-1
-- monday-2
-- saturday-7
-- Find the date of the nth future Sunday from a given date

-- Set your input values here
SELECT
  date(current_timestamp()) AS today_date,
  2 AS nth_sunday,
  date_add(
    date(current_timestamp()),
    (
      (8 - dayofweek(date(current_timestamp()))) % 7
      ) + (2 - 1) * 7
  ) AS nth_sunday_date
```
